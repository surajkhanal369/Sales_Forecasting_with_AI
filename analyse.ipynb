```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
# Load datasets
transactions = pd.read_csv("transaction_data.csv")
products = pd.read_csv("product.csv")
demographics = pd.read_csv("hh_demographic.csv")

print("Transaction Shape:", transactions.shape)
print("Product Shape:", products.shape)
print("Demographic Shape:", demographics.shape)
```

```python
def dataset_info(df, name):

    print("="*80)
    print(name.upper())
    print("="*80)

    print("\nFirst Five Rows")
    print(df.head())

    print("\nDataset Shape")
    print(df.shape)

    print("\nRows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nColumn Names")
    print(df.columns.tolist())

    print("\nData Types")
    print(df.dtypes)

    print("\nInformation")
    print(df.info())
```

```python
def missing_values(df):

    missing = pd.DataFrame({
        'Missing Values': df.isnull().sum(),
        'Percentage': round(df.isnull().mean()*100,2)
    })

    missing = missing.sort_values(
        by='Missing Values',
        ascending=False
    )

    return missing
```

```python
print(missing_values(transactions))

print(missing_values(products))

print(missing_values(demographics))
```

```python
def duplicate_info(df,name):

    print("="*60)

    print(name)

    print("Duplicate Rows:",df.duplicated().sum())

    print("Duplicate Percentage:",
          round(df.duplicated().mean()*100,2),"%")
```

```python
duplicate_info(transactions,"Transactions")

duplicate_info(products,"Products")

duplicate_info(demographics,"Demographics")
```

```python
def unique_values(df):

    unique = pd.DataFrame({
        "Unique Values":df.nunique()
    })

    return unique.sort_values(
        by="Unique Values",
        ascending=False
    )
```

```python
unique_values(transactions)

unique_values(products)

unique_values(demographics)
```

```python
def complete_summary(df,name):

    summary = pd.DataFrame({

        "Data Type":df.dtypes,

        "Missing":df.isnull().sum(),

        "Missing %":round(df.isnull().mean()*100,2),

        "Unique":df.nunique(),

        "Duplicates":df.duplicated().sum()

    })

    print("="*80)

    print(name)

    print("="*80)

    return summary
```

```python
complete_summary(transactions,"Transactions")
```

```python
complete_summary(products,"Products")
```

```python
complete_summary(demographics,"Demographics")
```
